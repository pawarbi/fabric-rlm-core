# Minimal RLM contract comparison

A tiny Fabric notebook that downloads two sample contracts, compares them with `fabric_rlm`, and returns the result payload.

```python
from urllib.request import urlretrieve

urls = [
    "https://raw.githubusercontent.com/Trampoline-AI/predict-rlm/2d93675d6d69b45f9eda9b8fc01e178323f8e6cb/examples/contract_comparison/sample/input/microFIT-Contract-Version-2-0.pdf",
    "https://raw.githubusercontent.com/Trampoline-AI/predict-rlm/2d93675d6d69b45f9eda9b8fc01e178323f8e6cb/examples/contract_comparison/sample/input/microFIT-Contract-Version-3-1-1.pdf",
]
pdf_paths = ["/tmp/contract_v2.pdf", "/tmp/contract_v3.pdf"]

for url, path in zip(urls, pdf_paths):
    urlretrieve(url, path)

pdf_paths
```

```python
from fabric_rlm import FabricLM, File, RLM

rlm = RLM.task(
    task="Compare the two contract PDFs. Return a concise summary and key_differences as a list of short strings.",
    inputs={"contracts": [File(path) for path in pdf_paths]},
    outputs=["summary", "key_differences"],
    lm=FabricLM("gpt-5.1", reasoning_effort="low"),
    skills=["pdf_document_analysis"],
    max_turns=8,
)

result = rlm.run()
result.payload
```
